Benchmark runs:
Setup:
- Benchmarks:
    - Image processing with 1 fan-out node (flip)
    - Text-2-Speech Censoring
    - Video analytics
- Invocation Rates:
    - Low invocation rate: 0.1 rps for 15 minutes (90 invocations)
    - High invocation rate: 1 rps for 5 minutes (300 invocations)
    - Very high invocation rate: 5 rps for 3 minutes (900 invocations)
- Concurrency Limits
    - 1 concurrency
    - 10 concurrency

- Measured latencies
    - End-to-end latency (from when the function is first invoked until the last function has finished execution)
    - Pub/sub latency (from when message is sent from sender until received - in this case, function starts execution)
    - Latencies will be served in stacked bar format, with the pub/sub latency stacked from each part of the function

- Amount of data being sent is dependent on the function that calls it (e.g. image processing and video analytics will have a different data size)
- Pub/sub settings
    - Topic settings
        - Schema
        - Ingestion (topic gets data asynchronously from a cloud storage bucket)
        - Message retention (up to 7 days)
        - Export message to BigQuery
        - Backup to google cloud storage
        - Message Transform
        - Encryption
    - Subscription settings
        - pull
        - push
        - write to BigQuery
        - write to Cloud storage
        - message retention
        - expiration period
        - ack deadline
        - transforms
        - subscription filter
        - exactly once delivery
        - ordering
        - dead lettering
        - retry policy
